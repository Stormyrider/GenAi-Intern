```python
import numpy as np
```

```python
numbers = [1, 2, 3, 4, 5]
arr = np.array(numbers)

print(numbers)
print(arr)
print(arr.dtype)
print(arr.shape)
```

```python
a = np.array([1, 2, 3, 4, 5])
zeros = np.zeros(5)
ones = np.ones(5)
numbers = np.arange(1, 6)
values = np.linspace(0, 10, 5)

print(a)
print(zeros)
print(ones)
print(numbers)
print(values)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[1:4])
print(arr[-1])
```

```python
arr = np.array([10, 25, 30, 45, 50])

print(arr[arr > 30])
```

```python
arr = np.array([1, 2, 3, 4, 5])

print(arr * 2)
print(arr + 10)
print(arr ** 2)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr.sum())
print(arr.mean())
print(arr.std())
print(arr.max())
print(arr.min())
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.vstack((a, b)))
print(np.hstack((a, b)))
```

```python
import time

numbers = list(range(1000000))

start = time.time()

result = []
for x in numbers:
    result.append(x ** 2)

python_time = time.time() - start

print("Python time:", python_time)
```

```python
numbers = np.arange(1000000)

start = time.time()

result = numbers ** 2

numpy_time = time.time() - start

print("NumPy time:", numpy_time)
```
