```python
paragraph = input("Enter a paragraph: ")

words = paragraph.lower().split()

count = {}

for word in words:
    if word in count:
        count[word] = count[word] + 1
    else:
        count[word] = 1

print("\nWord Frequency:")

for word in count:
    print(word, ":", count[word])
```
