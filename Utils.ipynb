```python
def add_contact(contacts: dict, name: str, phone: str) -> dict:
    """Add a contact to the contact book."""
    contacts[name] = phone
    return contacts


def search_contact(contacts: dict, name: str) -> str:
    """Search for a contact by name."""
    if name in contacts:
        return contacts[name]
    return "Contact not found."


def delete_contact(contacts: dict, name: str) -> dict:
    """Delete a contact from the contact book."""
    if name in contacts:
        del contacts[name]
    return contacts


def word_count(text: str) -> dict:
    """Count the frequency of each word."""
    words = text.lower().split()
    count = {}

    for word in words:
        if word in count:
            count[word] += 1
        else:
            count[word] = 1

    return count


def fizzbuzz(number: int) -> str:
    """Return the FizzBuzz result for a number."""
    if number % 3 == 0 and number % 5 == 0:
        return "FizzBuzz"
    elif number % 3 == 0:
        return "Fizz"
    elif number % 5 == 0:
        return "Buzz"
    return str(number)


def multiplication_table(number: int) -> None:
    """Print the multiplication table of a number."""
    for i in range(1, 11):
        print(number, "x", i, "=", number * i)


def add_numbers(a: int, b: int = 0) -> int:
    """Add two numbers."""
    return a + b


def average(*numbers: int) -> float:
    """Return the average of multiple numbers."""
    if len(numbers) == 0:
        return 0

    total = 0
    for num in numbers:
        total += num

    return total / len(numbers)


def student_info(**details) -> None:
    """Print student information."""
    for key in details:
        print(key, ":", details[key])
```

```python
contacts = {}

add_contact(contacts, "Ahmed", "03001234567")
print(search_contact(contacts, "Ahmed"))

print(word_count("Python is easy Python"))

print(fizzbuzz(15))

multiplication_table(5)

print(add_numbers(10, 20))

print(average(10, 20, 30, 40))

student_info(name="Ahmed", age=20, city="Karachi")
```
